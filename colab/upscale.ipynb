# 🔥 Sith Fireplace TV: AI upscale on a free Colab GPU

Runs **Real-ESRGAN x4plus** on every frame of the 60-second loop, using Colab's free T4 GPU (about 1 hour).

**Before you start:**
1. On your PC run `PREP_UPSCALE=1 ./build.sh`. It writes `work/upscale_input.mkv` (~145 MB).
2. Upload that file to Google Drive in a folder **`My Drive/sith_upscale/`**.
3. Here in Colab, choose **Runtime → Change runtime type → T4 GPU**.

Then run the cells in order (**Runtime → Run all**).
If Colab disconnects, run them all again: finished chunks are saved in Drive and skipped.

When it's done, download **`My Drive/sith_upscale/out/sith_upscaled.mp4`** to `work/sith_upscaled.mp4`
in the project and run `./build.sh` (add your TV address to install it).

```python
# 1. Check that a GPU is attached (it should list a Tesla T4)
!nvidia-smi --query-gpu=name,memory.total --format=csv
```

```python
# 2. Connect Google Drive (approve the pop-up)
from google.colab import drive
drive.mount('/content/drive')
import os
DIR = '/content/drive/MyDrive/sith_upscale'
SRC = f'{DIR}/upscale_input.mkv'
assert os.path.exists(SRC), f'Upload upscale_input.mkv to {DIR} first'
print('Found', SRC, round(os.path.getsize(SRC) / 1e6), 'MB')
```

```python
# 3. Get the upscaler script and the Real-ESRGAN x4plus weights (weights are cached in Drive)
!wget -q -O upscale.py https://raw.githubusercontent.com/crash101g/sith-fireplace-tv/main/colab/upscale.py
W = f'{DIR}/RealESRGAN_x4plus.pth'
if not os.path.exists(W):
    !wget -q -O "{W}" https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth
print('Weights:', round(os.path.getsize(W) / 1e6), 'MB')
```

```python
# 4. Upscale all 1,800 frames (progress prints after each 150-frame chunk)
!python upscale.py "{SRC}" "{DIR}/out" --weights "{W}"
```

Done! Download **`My Drive/sith_upscale/out/sith_upscaled.mp4`** from Google Drive and save it as
`work/sith_upscaled.mp4` in the project, then run `./build.sh`.
You can delete the `out/chunk_*.mp4` files afterwards.
